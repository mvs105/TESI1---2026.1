### Análise exploratória da Base de Dados do ENEM 2023

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import gc

pd.set_option('display.max_columns', None)

microdados_enem = pd.read_csv('base_dados_brutos/MICRODADOS_ENEM_2023_trabalhado.csv')
microdados_enem.head()
```

```python
pd.get_dummies(microdados_enem, columns=['TP_SEXO'])
```

```python
a = microdados_enem.isnull().sum()
print(a / len(microdados_enem)*100)
# microdados_enem.drop(inplace=True)
```

```python
dados_numericos = microdados_enem.select_dtypes(include='number').info(verbose=True)
print(dados_numericos)
```

```python
#Garbage Colector

gc.collect()
```

```python
copia = microdados_enem.copy()
indices = np.random.choice(copia.index, size=100, replace=False)
copia.loc[indices, 'TP_FAIXA_ETARIA'] = np.nan
sns.heatmap(copia.isnull())
plt.show()
removidos = copia.dropna()
print(removidos.shape)
```

```python
sns.histplot(data=microdados_enem, x='TP_FAIXA_ETARIA', bins=20)
plt.title('Distribuição da Faixa Etária dos participantes do ENEM')
plt.xlabel('Categoria')
plt.ylabel('Frequência')
```

```python
sns.barplot(data=microdados_enem, x='TP_SEXO', y='TP_ESCOLA', hue='TP_FAIXA_ETARIA')
plt.title('Gráfico de dispersão')
plt.ylabel('FAIXA ETARIA')
plt.xlabel('SEXO')
plt.show()
```

```python
#Transformação de variáveis númericas

from sklearn.preprocessing import StandardScaler, MinMaxScaler
total = microdados_enem['NU_NOTA_MT'].values.reshape(-1, 1)
tip = microdados_enem['NU_NOTA_CN'].values.reshape(-1, 1)
total_std = StandardScaler().fit_transform(total).ravel()
tip_nor = MinMaxScaler().fit_transform(tip).ravel()
copia['total_std'] = total_std
copia['tip_nor'] = tip_nor
copia.head()
```

```python
colunas = microdados_enem.select_dtypes(include=['float64'])
matriz = colunas.corr(method='pearson')
sns.heatmap(matriz)
```
