## Distribuição das notas de Matemática

O histograma permite visualizar como as notas estão distribuídas.

```python
from scipy import stats

print(df['NU_NOTA_MT'].skew())
print(df['NU_NOTA_MT'].kurt())

print(stats.shapiro(df['NU_NOTA_MT']))

plt.figure(figsize=(10,5))

sns.histplot(
    data=df,
    x="NU_NOTA_MT",
    bins=30,
    kde=True
)

plt.title("Distribuição das Notas de Matemática")
plt.xlabel("Nota")
plt.ylabel("Quantidade de candidatos")

plt.show()
```

```python
print(df['NU_NOTA_REDACAO'].skew())
print(df['NU_NOTA_REDACAO'].kurt())

print(stats.shapiro(df['NU_NOTA_REDACAO']))
plt.figure(figsize=(10,5))

sns.histplot(
    data=df,
    x="NU_NOTA_REDACAO",
    bins=30,
    kde=True,
    color="tomato"
)

plt.title("Distribuição das Notas da Redação")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="TP_SEXO"
)

plt.title("Distribuição por Sexo")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="TP_ESCOLA"
)

plt.title("Tipo de Escola")

plt.show()
```

```python
corr = df[notas].corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlação entre as notas")

plt.show()
```

```python
corr = df[notas].corr(method='spearman')

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlação entre as notas")

plt.show()
```
