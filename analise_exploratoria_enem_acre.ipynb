## Distribuição das notas de Matemática

O histograma permite visualizar como as notas estão distribuídas.

```python
from scipy import stats

print(df_acre['NU_NOTA_MT'].skew())
print(df_acre['NU_NOTA_MT'].kurt())

print(stats.shapiro(df_acre['NU_NOTA_MT']))

plt.figure(figsize=(10,5))

sns.histplot(
    data=df_acre,
    x="NU_NOTA_MT",
    bins=20,
    kde=True
)

plt.title("Distribuição das Notas de Matemática")
plt.xlabel("Nota")
plt.ylabel("Quantidade de candidatos")

plt.show()
```

```python
print(df_acre['NU_NOTA_REDACAO'].skew())
print(df_acre['NU_NOTA_REDACAO'].kurt())

print(stats.shapiro(df_acre['NU_NOTA_REDACAO']))
plt.figure(figsize=(10,5))

sns.histplot(
    data=df_acre,
    x="NU_NOTA_REDACAO",
    bins=30,
    kde=True,
    color="red"
)

plt.title("Distribuição das Notas da Redação")

plt.show()
```

```python
df_idades = (df_acre["TP_FAIXA_ETARIA"])
df_idades
sns.histplot(data=df_acre, x='TP_FAIXA_ETARIA', bins=20)
```

```python
plt.figure(figsize=(2,5))

sns.boxplot(
    y=df_idades,
    color="skyblue",
    width=0.35,
    showmeans=True,
    meanprops={"marker":"D", "markerfacecolor":"gold", "markeredgecolor":"black"},
    flierprops={"marker":"o", "markerfacecolor":"gray", "alpha":0.6, "markersize":4},
    medianprops={"color":"darkblue", "linewidth":2}
)

plt.title("Distribuição das Idades")
plt.ylabel("Idade")
plt.xticks([])  # esconder ticks do eixo x (apenas uma variável)
plt.ylim(df_idades.min() - 1, df_idades.max() + 1)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
df_idades = (df_acre["TP_FAIXA_ETARIA"] + 15).clip(lower=16, upper=35)
df_idades
sns.histplot(data=df_idades, bins=20)
```

```python
df_idades = (df_acre["TP_FAIXA_ETARIA"] + 15).clip(lower=16, upper=35)
df_idades
plt.figure(figsize=(10,5))

sns.histplot(
    data=df_idades,
    bins=20
)

plt.title("Distribuição das Idades")
plt.xlabel("Idade")
plt.ylabel("Quantidade de candidatos")
plt.xticks(sorted(df_idades.unique()))

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df_acre,
    x="TP_SEXO"
)

plt.title("Distribuição por Sexo")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df_acre,
    x="TP_ESCOLA"
)

plt.title("Tipo de Escola")

plt.show()
```

```python
corr = df_acre[notas].corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlação entre as notas")

plt.show()
```
