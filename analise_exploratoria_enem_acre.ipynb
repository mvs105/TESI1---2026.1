## Verificação de valores ausentes

Nesta etapa verificamos se existem valores faltantes na base.

A identificação desses registros permite avaliar a qualidade da base de dados e decidir se será necessário aplicar técnicas de tratamento ou remoção de informações incompletas.

```python
valores_nulos = df_acre.isnull().sum()

percentual = (df_acre.isnull().sum()/len(df_acre))*100

tabela_nulos = pd.DataFrame({
    "valores_nulos": valores_nulos.sort_values(ascending=False),
    "percentual": percentual.round(2).sort_values(ascending=False)
})

print(f"{'Coluna':<30} | {'Qtd. nulos':<10} | {'Percentual':<10}")
print("-" * 55)

# Mostra as 10 primiras colunas que mais tem valores nulos
for coluna, linha in tabela_nulos.head(20).iterrows():
    print(f"{coluna:<30} | {int(linha['valores_nulos']):<10} | {linha['percentual']:.2f} %")
```

### Distribuição por Faixa Etária

A variável **TP_FAIXA_ETARIA** é disponibilizada pelo INEP em formato codificado. Para facilitar a interpretação dos resultados, os códigos foram convertidos para suas respectivas descrições antes da construção do gráfico.

```python
df_idades = (df_acre["TP_FAIXA_ETARIA"])

mapeamento_faixas = {
    1: "Menos de 17",
    2: "17 anos",
    3: "18 anos",
    4: "19 anos",
    5: "20 anos",
    6: "21 anos",
    7: "22 anos",
    8: "23 anos",
    9: "24 anos",
    10: "25 anos",
    11: "26 até 30",
    12: "31 até 35",
    13: "36 até 40",
    14: "41 até 45",
    15: "46 até 50",
    18: "61 até 65",
    16: "51 até 55",
    17: "56 até 60",
    19: "66 até 70",
    20: "Maior de 70",
}

contagem = df_idades.value_counts().sort_index()

legendas_eixo_x = [mapeamento_faixas[codigo] for codigo in contagem.index]

plt.figure(figsize=(10, 6))

sns.barplot(
    x=legendas_eixo_x, y=contagem.values, palette="Blues_d", hue=legendas_eixo_x
)

plt.title("Distribuição por Faixa Etária", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Faixa Etária", fontsize=12)
plt.ylabel("Quantidade de Candidatos", fontsize=12)

plt.xticks(rotation=60, ha="right")

plt.show()
```

## Correlação entre as notas

O mapa de calor apresenta os coeficientes de correlação entre as notas das cinco áreas do ENEM.

Valores próximos de **1** indicam forte correlação positiva, enquanto valores próximos de **0** indicam pouca ou nenhuma relação linear entre as variáveis.

```python
colunas = df_acre[notas]
matriz = colunas.corr(method='pearson')

plt.figure(figsize=(8,6))

nomes_labels = ['CN', 'CH', 'LC', 'MT', 'RED']

sns.heatmap(
    matriz,
    annot=True,
    cmap="coolwarm",
    vmin=0.4, vmax=1,
    fmt=".2f",
    xticklabels=nomes_labels,
    yticklabels=nomes_labels 
)

plt.title("Correlação entre as notas" , fontsize=16, fontweight="bold", pad=15)

plt.show()
```

## Relação entre as notas de Matemática e Linguagens - Gráfico de dispersão

Cada ponto representa um participante do ENEM, e a distribuição está categorizada pelo tipo de administração da escola.

```python
df_acre_filtrado = df_acre[(df_acre["NU_NOTA_MT"] > 0) & (df_acre["NU_NOTA_LC"] > 0)]

cores = {
    1: "orange",
    2: "green",
    3: "red",
    4: "blue"
}

sns.scatterplot(
    data=df_acre_filtrado, 
    x='NU_NOTA_MT',
    y='NU_NOTA_LC', 
    alpha=0.7, 
    hue='TP_DEPENDENCIA_ADM_ESC',
    palette=cores,
)

plt.title("Notas de Matemática x Linguagens", fontsize=16, fontweight="bold", pad=15)
plt.xlabel('Notas de Matemática')
plt.ylabel('Notas de Linguagens e Códigos')
plt.legend(title = 'Dependência administrativa', labels = ['Federal', 'Estadual', 'Municipal', 'Privada'])

plt.show()
```
