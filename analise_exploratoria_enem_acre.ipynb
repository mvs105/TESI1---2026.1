## Distribuição das notas de Matemática

O histograma permite visualizar como as notas estão distribuídas.

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df_acre[
        [
            "NU_NOTA_CN",
            "NU_NOTA_CH",
            "NU_NOTA_LC",
            "NU_NOTA_MT",
            "NU_NOTA_REDACAO"
        ]
    ]
)

plt.title("Distribuição das notas")
```

```python
from scipy import stats

# print(df_acre['NU_NOTA_MT'].skew())
# print(df_acre['NU_NOTA_MT'].kurt())

# print(stats.shapiro(df_acre['NU_NOTA_MT']))

plt.figure(figsize=(10,5))

sns.histplot(
    data=df_acre,
    x="NU_NOTA_MT",
    bins=20,
    kde=True
)

plt.title("Distribuição das Notas de Matemática")
plt.xlabel("Nota")
plt.ylabel("Quantidade de candidatos")

plt.show()
```

```python
# print(df_acre['NU_NOTA_REDACAO'].skew())
# print(df_acre['NU_NOTA_REDACAO'].kurt())

# print(stats.shapiro(df_acre['NU_NOTA_REDACAO']))
plt.figure(figsize=(10,5))

sns.histplot(
    data=df_acre,
    x="NU_NOTA_REDACAO",
    bins=30,
    kde=True,
    color="red"
)

plt.title("Distribuição das Notas da Redação")

plt.show()
```

```python
df_idades = (df_acre["TP_FAIXA_ETARIA"])

mapeamento_faixas = {
    1: "Menos de 17",
    2: "17 anos",
    3: "18 anos",
    4: "19 anos",
    5: "20 anos",
    6: "21 anos",
    7: "22 anos",
    8: "23 anos",
    9: "24 anos",
    10: "25 anos",
    11: "26 até 30",
    12: "31 até 35",
    13: "36 até 40",
    14: "41 até 45",
    15: "46 até 50",
    16: "51 até 55",
    17: "56 até 60",
    18: "61 até 65",
    19: "66 até 70",
    20: "Maior de 70",
}

contagem = df_idades.value_counts().sort_index()

legendas_eixo_x = [mapeamento_faixas[codigo] for codigo in contagem.index]

plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid") 

grafico = sns.barplot(
    x=legendas_eixo_x, y=contagem.values, palette="Blues_d", hue=legendas_eixo_x
)

plt.title("Distribuição por Faixa Etária", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Faixa Etária", fontsize=12)
plt.ylabel("Quantidade de Pessoas", fontsize=12)

plt.xticks(rotation=60, ha="right")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df_acre,
    x="TP_SEXO"
)

plt.title("Distribuição por Sexo")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df_acre,
    x="TP_ESCOLA"
)

plt.title("Tipo de Escola")

plt.show()
```

```python
corr = df_acre[notas].corr()

plt.figure(figsize=(8,6))

nomes_labels = ['CN', 'CH', 'LC', 'MT', 'RED']

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    xticklabels=nomes_labels,
    yticklabels=nomes_labels 
)

plt.title("Correlação entre as notas")

plt.show()
```

```python
df_acre_filtrado = df_acre[(df_acre["NU_NOTA_MT"] > 0) & (df_acre["NU_NOTA_LC"] > 0)]

minhas_cores = {
    1: "orange",
    2: "green",
    3: "red",
    4: "blue"
}

sns.scatterplot(
    data=df_acre_filtrado, 
    x='NU_NOTA_MT',
    y='NU_NOTA_LC', 
    alpha=1, 
    hue='TP_DEPENDENCIA_ADM_ESC',
    palette=minhas_cores,
)

plt.title("Notas de Matemática x Linguagens")
plt.xlabel('Notas de Matemática')
plt.ylabel('Notas de Linguagens e Códigos')
plt.legend(title = 'Dependência administrativa', labels = ['Federal', 'Estadual', 'Municipal', 'Privada'])

plt.show()
```
